```python
# TEXT EMBEDDINGS PROJECT

import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
# Load the sentence dataset

df = pd.read_csv("Data/text_generation_results.csv")

print("Dataset loaded successfully!")
print("Number of sentences:", len(df))

df.head()
```

```python
# Get sentences from the dataset

sentences = df["sentence"].tolist()

print("Sentences extracted successfully!")
print("Total sentences:", len(sentences))
```

```python
# Load the embedding model

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
# Generate embeddings

embeddings = model.encode(sentences)

print("Embeddings generated successfully!")
print("Embedding shape:", embeddings.shape)
```

```python
# Save embeddings

np.save("Data/embeddings.npy", embeddings)

print("Embeddings saved successfully!")
```

```python
# Calculate cosine similarity

similarity_matrix = cosine_similarity(embeddings)

print("Cosine similarity calculated successfully!")
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
# Create sentence pairs with similarity scores

results = []

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        results.append({
            "Text 1": sentences[i],
            "Text 2": sentences[j],
            "Cosine Similarity": round(similarity_matrix[i][j], 4)
        })

results_df = pd.DataFrame(results)

print("Sentence pairs created:", len(results_df))
results_df.head()
```

```python
# Sort pairs by similarity

results_df = results_df.sort_values(
    by="Cosine Similarity",
    ascending=False
)

results_df.head(5)
```

```python
# Save all similarity results

results_df.to_csv(
    "Data/similarity_results.csv",
    index=False
)

print("Similarity results saved successfully!")
```

```python
# Save top 10 most similar pairs

top_10 = results_df.head(10)

top_10.to_csv(
    "Data/top_10_pairs.csv",
    index=False
)

print("Top 10 pairs saved successfully!")
```

```python
# Display Top 5 Most Similar Sentence Pairs

print("TOP 5 MOST SIMILAR SENTENCE PAIRS\n")

top_5 = results_df.head(5)

for index, row in top_5.iterrows():
    print("Text 1:", row["Text 1"])
    print("Text 2:", row["Text 2"])
    print("Cosine Similarity:", row["Cosine Similarity"])
    print("-" * 80)
```

```python
# Final Project Summary

print("TEXT EMBEDDINGS PROJECT")
print("-----------------------")
print("Total sentences:", len(sentences))
print("Embedding dimensions:", embeddings.shape[1])
print("Total sentence pairs compared:", len(results_df))
print("Top 5 pairs identified: Yes")
print("Embeddings saved: Yes")
print("Similarity results saved: Yes")
print("Project completed successfully!")
```

# TEXT EMBEDDINGS PROJECT

## Objective
To generate text embeddings for 20 sentences and compare their semantic similarity using cosine similarity.

## Methodology

1. Collected 20 sentences from different topics.
2. Loaded the Sentence Transformer model.
3. Converted the sentences into 384-dimensional embeddings.
4. Calculated cosine similarity between sentence pairs.
5. Compared 190 unique sentence pairs.
6. Identified the top 5 most similar pairs.
7. Saved embeddings and similarity results as output files.

## Results

The embeddings were generated successfully for all 20 sentences. A total of 190 unique sentence pairs were compared using cosine similarity.

### Top 5 Most Similar Pairs

The top five pairs were identified based on their cosine similarity scores. Higher scores indicate greater semantic similarity.

## Observations

- Sentence embeddings represent the semantic meaning of text as numerical vectors.
- Related topics generally produced higher cosine similarity scores.
- Space-related and environment-related sentences showed relatively high similarity.
- Unrelated topics generally produced lower similarity scores.
- Embeddings can identify meaning even when the exact words are different.

## Conclusion

This project demonstrates how text embeddings can be used to represent sentence meaning and measure semantic similarity using cosine similarity. Text embeddings are useful in semantic search, recommendation systems, document retrieval and LLM applications.
